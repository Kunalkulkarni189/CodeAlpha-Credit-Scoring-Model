```python
# Input features
X = df.drop("class", axis=1)

# Target variable
y = df["class"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numerical_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_columns = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_columns),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_columns)
    ]
)

print("Numerical features:", numerical_columns)
print("Categorical features:", categorical_columns)
```

```python
print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.pipeline import Pipeline

print("All ML algorithms imported successfully!")
```

```python
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42
    )
}
```

```python
pipelines = {}

for name, model in models.items():

    pipelines[name] = Pipeline([
        ("preprocessing", preprocessor),
        ("classifier", model)
    ])

print("All three pipelines created successfully!")
```

```python
for name, pipeline in pipelines.items():

    print(f"Training {name}...")

    pipeline.fit(X_train, y_train)

    print(f"{name} trained successfully!\n")
```

```python
predictions = {}

for name, pipeline in pipelines.items():

    y_pred = pipeline.predict(X_test)

    predictions[name] = y_pred

    print(f"\n{name}")
    print("First 15Predictions:")
    print(y_pred[:15])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

results = []

for name, pipeline in pipelines.items():

    y_pred = pipeline.predict(X_test)

    y_prob = pipeline.predict_proba(X_test)

    bad_index = list(pipeline.classes_).index("Bad")

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred,
            pos_label="Bad",
            zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred,
            pos_label="Bad",
            zero_division=0
        ),
        "F1 Score": f1_score(
            y_test, y_pred,
            pos_label="Bad",
            zero_division=0
        ),
        "ROC-AUC": roc_auc_score(
            (y_test == "Bad").astype(int),
            y_prob[:, bad_index]
        )
    })

results_df = pd.DataFrame(results)

print(results_df.round(3))
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.metrics import roc_auc_score
```

```python
def bad_roc_auc(estimator, X_val, y_val):
    bad_index = list(estimator.classes_).index("Bad")
    bad_probs = estimator.predict_proba(X_val)[:, bad_index]

    y_true = (y_val == "Bad").astype(int)

    return roc_auc_score(y_true, bad_probs)
```

```python
param_grids = {
    "Logistic Regression": {
        "classifier__C": [0.1, 1, 10],
        "classifier__class_weight": [None, "balanced"]
    },

    "Decision Tree": {
        "classifier__max_depth": [3, 5, None],
        "classifier__min_samples_leaf": [1, 5],
        "classifier__class_weight": [None, "balanced"]
    },

    "Random Forest": {
        "classifier__n_estimators": [100, 200],
        "classifier__max_depth": [None, 10],
        "classifier__min_samples_leaf": [1, 3],
        "classifier__class_weight": [None, "balanced"]
    }
}
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

best_pipelines = {}
tuning_results = []

for name, pipeline in pipelines.items():

    grid = GridSearchCV(
        estimator=pipeline,
        param_grid=param_grids[name],
        scoring=bad_roc_auc,
        cv=cv,
        n_jobs=-1,
        refit=True
    )

    grid.fit(X_train, y_train)

    best_pipelines[name] = grid.best_estimator_

    tuning_results.append({
        "Model": name,
        "Best CV ROC-AUC": grid.best_score_,
        "Best Parameters": grid.best_params_
    })

    print("\nModel:", name)
    print("Best ROC-AUC:", round(grid.best_score_, 3))
    print("Best Parameters:", grid.best_params_)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

tuned_results = []

for name, pipeline in best_pipelines.items():

    y_pred = pipeline.predict(X_test)
    y_prob = pipeline.predict_proba(X_test)

    bad_index = list(pipeline.classes_).index("Bad")

    tuned_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred, pos_label="Bad", zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, pos_label="Bad", zero_division=0
        ),
        "F1 Score": f1_score(
            y_test, y_pred, pos_label="Bad", zero_division=0
        ),
        "ROC-AUC": roc_auc_score(
            (y_test == "Bad").astype(int),
            y_prob[:, bad_index]
        )
    })

tuned_results_df = pd.DataFrame(tuned_results)

print(tuned_results_df.round(3))
```
